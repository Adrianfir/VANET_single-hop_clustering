```python
import numpy as pd
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/Users/pouyafirouzmakan/Desktop/all.csv')
```

```python
sns.set(rc={
    "font.family": "serif",
    "font.serif": ["Times New Roman"],
    "mathtext.fontset": "stix",
    "mathtext.default": "it",
    "text.usetex": False
}, style='whitegrid')

custome_palette = {'no':'0.75', 
                   'yes':'b'}
custom_boxprops = {'edgecolor': "#137"}
####################################

fig, axes = plt.subplots(nrows=1,ncols=3, figsize=(10,4), dpi=500)
x_order = ["no", "yes"]

sns.boxplot(data = df[df['TR']==100], y='stab_eval', x='rsu',notch=False, ax=axes[0], palette=custome_palette, boxprops=custom_boxprops, whiskerprops={'color': '#137'}, capprops={'color': '#137'}, medianprops={'color': '#137'},linewidth=1, order=x_order)
axes[0].grid('both')
# axes[0][0].set_title('TR = 100m')
axes[0].set(xlabel='case study', ylabel='VCSM')
axes[0].set_xticklabels(['1', '2']) 

sns.boxplot(data = df[df['TR']==200], y='stab_eval', x='rsu',notch=False, ax=axes[1], palette=custome_palette, boxprops=custom_boxprops, whiskerprops={'color': '#137'}, capprops={'color': '#137'}, medianprops={'color': '#137'},linewidth=1, order=x_order)
axes[1].grid('both')
# axes[1].set_title('TR = 200m')
axes[1].set(xlabel='case study', ylabel='VCSM')
axes[1].set_xticklabels(['1', '2'])

sns.boxplot(data = df[df['TR']==300], y='stab_eval', x='rsu',notch=False, ax=axes[2], palette=custome_palette, boxprops=custom_boxprops, whiskerprops={'color': '#137'}, capprops={'color': '#137'}, medianprops={'color': '#137'},linewidth=1, order=x_order)
axes[2].grid('both')
# axes[1][0].set_title('TR = 300m')
axes[2].set(xlabel='case study', ylabel='VCSM')
axes[2].set_xticklabels(['1', '2'])

subplots_labels = ['(a)', '(b)', '(c)', '(d)']
for i, ax in enumerate(axes.flatten()):
    ax.text(0.465, -0.25, subplots_labels[i], transform=ax.transAxes, fontsize=16, va='bottom')

plt.tight_layout()

plt.savefig('/Users/pouyafirouzmakan/Desktop/boxplots.png')
```

```python
import matplotlib.patches as mpatches

sns.set(rc={
    "font.family": "serif",
    "font.serif": ["Times New Roman"],
    "font.size": 20,
    "mathtext.fontset": "stix",
    "mathtext.default": "it",
    "text.usetex": False
}, style='whitegrid')
custome_palette = {'no': '0.75', 'yes': 'b'}
custom_boxprops = {'edgecolor': "#137"}
custom_flierprops = {'marker': 'D', 'color': 'b', 'markerfacecolor': 'k', 'markersize': 3}
hue_order = ['no', 'yes']
plt.figure(figsize=(4.5, 4), dpi=500)

# Boxplot with custom outliers
sns.boxplot(data=df[df['TR'].isin([100, 200, 300])], y='stab_eval', x='TR', hue='rsu', notch=False, 
            palette=custome_palette, whiskerprops={'color': '#137'}, capprops={'color': '#137'}, 
            medianprops={'color': '#137'}, linewidth=1, boxprops=custom_boxprops, 
            flierprops=custom_flierprops, hue_order=hue_order)

# Create custom legend handles with color matching the boxes
legend_handles = [
    mpatches.Patch(color='0.75', label='Case Study 1'),
    mpatches.Patch(color='b', label='Case Study 2')
]

# Add the custom legend
plt.legend(handles=legend_handles)
plt.xlabel('transmission range (m)')
plt.ylabel('VCSM')
plt.grid('both')

plt.tight_layout()
plt.savefig('/Users/pouyafirouzmakan/Desktop/eval_boxplots.png')
```
