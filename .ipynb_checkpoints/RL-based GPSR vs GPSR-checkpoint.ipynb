```python
import seaborn as sns
import matplotlib.pyplot as plt

# ----------------------------------------
# STYLE
# ----------------------------------------
sns.set(rc={
    "font.family": "serif",
    "font.serif": ["Times New Roman"],
    "mathtext.fontset": "stix",
    "mathtext.default": "it",
    "text.usetex": False
}, style='whitegrid')

custom_palette = {
    'ZRL-GPSR': '0.50',
    'GPSR': 'b'
}

# ----------------------------------------
# DATA FILTER (TEST DATA ONLY)
# ----------------------------------------
small_test = df[df['data_type'] == 'Small Data, TR=300m (test_data)']
medium_test = df[df['data_type'] == 'Medium Data & Medium Message data, TR=300m']

df_plot = pd.concat([small_test, medium_test], ignore_index=True)
df_plot['sparsity'] = df_plot['data_type'].apply(
    lambda x: "High Sparsity" if "Small Data" in x else "Medium Sparsity"
)

x_order = ["High Sparsity", "Medium Sparsity"]
hue_order = ["ZRL-GPSR", "GPSR"]

# ----------------------------------------
# FIGURE
# ----------------------------------------
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(10, 8), dpi=500)

########################################
# 1 — Delivery Ratio
########################################
sns.barplot(data=df_plot,
            x='sparsity',
            y='del_ratio',
            hue='protocol',
            ax=axes[0,0],
            # palette=custom_palette,
            order=x_order,
            hue_order=hue_order,
            edgecolor='black')
axes[0,0].grid('both')
axes[0,0].set(xlabel='Sparsity Level', ylabel='Delivery Ratio')
axes[0,0].set_xticklabels(['High Sparsity', 'Medium Sparsity'])
axes[0,0].set_ylim(0, 100)
axes[0,0].get_legend().remove()

########################################
# 2 — Average Hops
########################################
sns.barplot(data=df_plot,
            x='sparsity',
            y='avg_hops',
            hue='protocol',
            ax=axes[0,1],
            # palette=custom_palette,
            order=x_order,
            hue_order=hue_order,
            edgecolor='black')
axes[0,1].grid('both')
axes[0,1].set(xlabel='Sparsity Level', ylabel='Average Hops')
axes[0,1].set_xticklabels(['High Sparsity', 'Medium Sparsity'])
axes[0,1].get_legend().remove()

########################################
# 3 — Delay
########################################
sns.barplot(data=df_plot,
            x='sparsity',
            y='delay',
            hue='protocol',
            ax=axes[1,0],
            # palette=custom_palette,
            order=x_order,
            hue_order=hue_order,
            edgecolor='black')
axes[1,0].grid('both')
axes[1,0].set(xlabel='Sparsity Level', ylabel='Delay')
axes[1,0].set_xticklabels(['High Sparsity', 'Medium Sparsity'])
axes[1,0].get_legend().remove()

########################################
# 4 — Number of Perimeter Hops
########################################
sns.barplot(data=df_plot,
            x='sparsity',
            y='num_perimeter',
            hue='protocol',
            ax=axes[1,1],
            # palette=custom_palette,
            order=x_order,
            hue_order=hue_order,
            edgecolor='black')
axes[1,1].grid('both')
axes[1,1].set(xlabel='Sparsity Level', ylabel='Number of Perimeter Hops')
axes[1,1].set_xticklabels(['High Sparsity', 'Medium Sparsity'])
axes[1,1].get_legend().remove()

########################################
# subplot labels
########################################
subplots_labels = ['(a)', '(b)', '(c)', '(d)']
axes[0,0].text(0.48, -0.2, subplots_labels[0], transform=axes[0,0].transAxes, fontsize=14, va='top')
axes[0,1].text(0.48, -0.2, subplots_labels[1], transform=axes[0,1].transAxes, fontsize=14, va='top')
axes[1,0].text(0.48, -0.2, subplots_labels[2], transform=axes[1,0].transAxes, fontsize=14, va='top')
axes[1,1].text(0.48, -0.2, subplots_labels[3], transform=axes[1,1].transAxes, fontsize=14, va='top')

########################################
# Legend once for whole fig
########################################
handles, labels = axes[0,0].get_legend_handles_labels()
fig.legend(handles, labels, loc='upper center', ncol=3, frameon=True)

plt.tight_layout()

# SAVE FIGURE (optional)
# plt.savefig('/Users/.../barplots_small_medium_test.png', dpi=500, bbox_inches='tight')

plt.show()
```
